````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
import sqlite3

conn = sqlite3.connect("activity.db")
cursor = conn.cursor()

try:
    # 先刪除舊資料表
    cursor.execute("DROP TABLE IF EXISTS registrations")
    cursor.execute("DROP TABLE IF EXISTS activities")

    # 建立活動資料表
    cursor.execute("""
        CREATE TABLE activities (
            activity_id INTEGER PRIMARY KEY AUTOINCREMENT,
            activity_name TEXT UNIQUE,
            activity_date TEXT,
            capacity INTEGER
        )
    """)

    # 建立報名資料表
    cursor.execute("""
        CREATE TABLE registrations (
            registration_id INTEGER PRIMARY KEY AUTOINCREMENT,
            student_name TEXT,
            activity_id INTEGER,
            FOREIGN KEY (activity_id)
            REFERENCES activities(activity_id)
        )
    """)

    # 初始活動資料
    activities_data = [
        ("迎新晚會", "2026-11-20", 6),
        ("程式工作坊", "2026-11-25", 4),
        ("籃球友誼賽", "2026-11-28", 3)
    ]

    cursor.executemany("""
        INSERT INTO activities
        (activity_name, activity_date, capacity)
        VALUES (?, ?, ?)
    """, activities_data)

    # 初始報名資料
    initial_registrations = [
        ("初始學生1", "迎新晚會"),
        ("初始學生2", "迎新晚會"),
        ("初始學生3", "迎新晚會"),
        ("初始學生4", "迎新晚會"),
        ("初始學生5", "迎新晚會"),

        ("初始學生6", "程式工作坊"),
        ("初始學生7", "程式工作坊"),

        ("初始學生8", "籃球友誼賽"),
        ("初始學生9", "籃球友誼賽"),
        ("初始學生10", "籃球友誼賽")
    ]

    for student, activity in initial_registrations:
        cursor.execute("""
            SELECT activity_id
            FROM activities
            WHERE activity_name = ?
        """, (activity,))

        result = cursor.fetchone()

        if result is not None:
            cursor.execute("""
                INSERT INTO registrations
                (student_name, activity_id)
                VALUES (?, ?)
            """, (student, result[0]))

    conn.commit()

    print("資料庫建立完成！")
    print("活動資料及初始報名資料已新增。")

except Exception as e:
    conn.rollback()
    print("資料庫建立失敗：", e)

finally:
    conn.close()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
import sqlite3
import pandas as pd
from IPython.display import display

def show_tables():
    conn = sqlite3.connect("activity.db")
    try:
        # 使用 JOIN 和 GROUP BY 統計資料
        sql = """
            SELECT
                a.activity_name AS 活動,
                a.activity_date AS 日期,
                a.capacity AS 總名額,
                COUNT(r.registration_id) AS 報名人數,
                a.capacity - COUNT(r.registration_id) AS 剩餘名額
            FROM activities a
            LEFT JOIN registrations r
            ON a.activity_id = r.activity_id
            GROUP BY
                a.activity_id,
                a.activity_name,
                a.activity_date,
                a.capacity
            ORDER BY a.activity_id
        """

        df = pd.read_sql_query(sql, conn)
        print("目前活動資料：")
        display(df)

    except Exception as e:
        print("查詢資料失敗：", e)
    finally:
        conn.close()

# 呼叫一次顯示目前資料表
show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================

# 允許的活動名稱
ALLOWED_ACTIVITIES = [
    "迎新晚會",
    "程式工作坊",
    "籃球友誼賽"
]

# 允許的動作
ALLOWED_ACTIONS = [
    "報名成功",
    "取消成功",
    "顯示活動名稱",
    "顯示剩餘名額",
    "查無條件"
]

# AI 解讀結果快取
ai_cache = {}

def ai_interpret(user_text):

    # 如果相同句子已經解讀過，直接使用快取
    if user_text in ai_cache:
        return ai_cache[user_text]

    prompt = f"""
你是一個系學會活動報名系統的語意解讀助手。

請判斷使用者的話，輸出以下格式：
動作,活動,人數

允許的動作：
- 報名成功：使用者想報名
- 取消成功：使用者想取消報名
- 顯示活動名稱：使用者想查詢有哪些活動
- 顯示剩餘名額：使用者想查詢剩餘名額
- 查無條件：無法判斷或與活動無關

允許的活動：
- 迎新晚會
- 程式工作坊
- 籃球友誼賽

規則：
1. 不可以創造新活動。
2. 活動名稱必須使用允許清單中的名稱。
3. 如果使用者只說籃球賽，代表籃球友誼賽。
4. 查詢所有活動時，活動填寫全部。
5. 無法判斷時輸出查無條件。
6. 不要輸出其他解釋文字。
7. 人數請依照使用者語意提供，沒有提到就填1。

使用者說：
{user_text}

請輸出：
"""

    raw_response = ""
    max_retries = 3

    for attempt in range(max_retries):

        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=[
                    {
                        "role": "user",
                        "content": prompt
                    }
                ],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )

            time.sleep(2)

            raw_response = response.choices[0].message.content.strip()
            break

        except Exception as e:

            error_message = str(e)

            if "429" in error_message:

                print(f"使用量超過限制，等待30秒後重試（第 {attempt + 1} 次）")

                if attempt < max_retries - 1:
                    time.sleep(30)
                else:
                    raw_response = "查無條件,未知,1"

            else:
                print("AI 呼叫失敗：", e)
                raw_response = "查無條件,未知,1"
                break

    # 從 AI 原始回覆中擷取允許的動作
    action = "查無條件"

    for allowed in ALLOWED_ACTIONS:
        if allowed in raw_response:
            action = allowed
            break

    # 從 AI 原始回覆中擷取允許的活動
    activity = "未知"

    for allowed in ALLOWED_ACTIVITIES:
        if allowed in raw_response:
            activity = allowed
            break

    # 查詢全部活動
    if "全部" in raw_response and action in [
        "顯示活動名稱",
        "顯示剩餘名額"
    ]:
        activity = "全部"

    # 使用正規表示式擷取數量
    number_match = re.search(r"\d+", raw_response)

    if number_match:
        quantity = int(number_match.group())
    else:
        quantity = 1

    # 數量必須大於0
    if quantity <= 0:
        quantity = 1

    # 如果活動不在允許清單中，視為未知
    if activity not in ALLOWED_ACTIVITIES and activity != "全部":
        activity = "未知"
        action = "查無條件"

    # 如果動作不在允許清單中，視為查無條件
    if action not in ALLOWED_ACTIONS:
        action = "查無條件"

    result = {
        "raw": raw_response,
        "action": action,
        "activity": activity,
        "quantity": quantity
    }

    # 儲存快取
    ai_cache[user_text] = result

    return result
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
import re

ai_tests = [
    ("我想報名籃球賽", "報名成功", "籃球友誼賽", 1),
    ("我想取消報名籃球賽", "取消成功", "籃球友誼賽", 1),
    ("我想報名程式工作坊", "報名成功", "程式工作坊", 1),
    ("我想取消報名程式工作坊", "取消成功", "程式工作坊", 1),
    ("我想報名迎新晚會", "報名成功", "迎新晚會", 1),
    ("我想取消報名迎新晚會", "取消成功", "迎新晚會", 1),
    ("我想查詢剩餘活動名額", "顯示剩餘名額", "全部", 1),
    ("我想報名畫畫比賽", "查無條件", "未知", 1),
    ("我想報名書法比賽", "查無條件", "未知", 1)
]

passed = 0

# 確保 re 模組在全域也正確被載入，防止第 4 格報錯
import sys
import sys as _sys
_sys.modules['re'] = re
globals()['re'] = re

try:
    for i, test in enumerate(ai_tests):
        user_text, expected_action, expected_activity, expected_quantity = test
        result = ai_interpret(user_text)

        print("=" * 60)
        print(f"測試第 {i + 1} 句")
        print("使用者：", user_text)
        print("AI 原始回覆：")
        print(result["raw"])
        print("系統解讀：")
        print(result["action"], result["activity"], result["quantity"])

        # 比較測試結果
        checks = [
            expected_action == "*" or result["action"] == expected_action,
            expected_activity == "*" or result["activity"] == expected_activity,
            expected_quantity == "*" or result["quantity"] == expected_quantity
        ]

        if all(checks):
            print("<測試通過>")
            passed += 1
        else:
            print("<測試未通過>")

    print("=" * 60)
    print(f"AI 解讀測試通過：{passed}/{len(ai_tests)} 句")

except NameError as ne:
    if "client" in str(ne):
        print("❌ 錯誤：找不到 'client'。請確認您已經執行過最上方的『第 1 格』以初始化 Groq 客戶端！")
    else:
        print("發生錯誤：", ne)
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================

def handle_request(student_name, user_text):

    # 先進行 AI 解讀
    result = ai_interpret(user_text)

    action = result["action"]
    activity = result["activity"]
    quantity = result["quantity"]

    # 數量範圍檢查
    if quantity < 1 or quantity > 100:
        return "數量超出允許範圍。"

    # 無法判斷
    if action == "查無條件":
        return "查無條件，請輸入與活動報名有關的內容。"

    conn = sqlite3.connect("activity.db")
    cursor = conn.cursor()

    try:

        # 查詢活動名稱
        if action == "顯示活動名稱":

            cursor.execute("""
                SELECT activity_name, activity_date
                FROM activities
                ORDER BY activity_id
            """)

            rows = cursor.fetchall()

            if not rows:
                return "目前沒有活動。"

            message = "目前活動：\n"

            for row in rows:
                message += f"{row[0]}（{row[1]}）\n"

            return message

        # 查詢剩餘名額
        if action == "顯示剩餘名額":

            if activity == "全部":

                cursor.execute("""
                    SELECT
                        a.activity_name,
                        a.capacity - COUNT(r.registration_id)
                    FROM activities a
                    LEFT JOIN registrations r
                    ON a.activity_id = r.activity_id
                    GROUP BY a.activity_id
                    ORDER BY a.activity_id
                """)

                rows = cursor.fetchall()

                if not rows:
                    return "目前沒有活動。"

                message = "各活動剩餘名額：\n"

                for row in rows:
                    message += f"{row[0]}：剩餘 {row[1]} 人\n"

                return message

            cursor.execute("""
                SELECT
                    a.activity_name,
                    a.capacity - COUNT(r.registration_id)
                FROM activities a
                LEFT JOIN registrations r
                ON a.activity_id = r.activity_id
                WHERE a.activity_name = ?
                GROUP BY a.activity_id
            """, (activity,))

            row = cursor.fetchone()

            if row is None:
                return "查無此活動。"

            return f"{row[0]} 剩餘名額：{row[1]} 人"

        # 其他功能必須指定單一活動
        if activity not in ALLOWED_ACTIVITIES:
            return "查無此活動。"

        # 查詢活動資料
        cursor.execute("""
            SELECT activity_id, capacity
            FROM activities
            WHERE activity_name = ?
        """, (activity,))

        activity_data = cursor.fetchone()

        if activity_data is None:
            return "查無此活動。"

        activity_id = activity_data[0]
        capacity = activity_data[1]

        # 查詢目前報名人數
        cursor.execute("""
            SELECT COUNT(*)
            FROM registrations
            WHERE activity_id = ?
        """, (activity_id,))

        count_data = cursor.fetchone()

        if count_data is None:
            return "無法取得報名資料。"

        registered_count = count_data[0]

        # 報名功能
        if action == "報名成功":

            # 檢查是否有足夠名額
            if registered_count + quantity > capacity:
                return (
                    f"{activity} 名額不足！"
                    f"目前剩餘 {capacity - registered_count} 人。"
                )

            # 新增報名紀錄
            for i in range(quantity):

                cursor.execute("""
                    INSERT INTO registrations
                    (student_name, activity_id)
                    VALUES (?, ?)
                """, (student_name, activity_id))

            conn.commit()

            return f"{student_name}，{activity} 報名成功！"

        # 取消報名功能
        elif action == "取消成功":

            # 確認使用者是否有報名
            cursor.execute("""
                SELECT registration_id
                FROM registrations
                WHERE student_name = ?
                AND activity_id = ?
                ORDER BY registration_id
                LIMIT ?
            """, (student_name, activity_id, quantity))

            registrations = cursor.fetchall()

            # 必須有足夠的報名紀錄才能取消
            if len(registrations) < quantity:
                return "取消失敗，你沒有足夠的報名紀錄。"

            # 刪除指定報名紀錄
            for registration in registrations:

                cursor.execute("""
                    DELETE FROM registrations
                    WHERE registration_id = ?
                """, (registration[0],))

            conn.commit()

            return f"{student_name}，{activity} 取消成功！"

        return "查無條件。"

    except Exception as e:

        conn.rollback()
        return f"系統發生錯誤：{e}"

    finally:
        conn.close()
````
